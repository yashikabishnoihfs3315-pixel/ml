Question 1: Coefficients, Intercept and House Price Prediction

```python

# Multiple Linear Regression: House Price Prediction

def predict_price(area, bedrooms, age):
    price = 15 + 0.08 * area + 5 * bedrooms - 0.6 * age
    return price

# House details
houses = {"A": (1500, 3, 5),
    "B": (2200, 4, 10),
    "C": (1000, 2, 15)}

# Predict prices
for house, (area, bedrooms, age) in houses.items():
    price = predict_price(area, bedrooms, age)
    print(f"House {house}: ₹{price:.2f} lakhs")
```

Question 2: Integrated Regression Model Challenge

```python

# Used-Car Price Prediction

def predict_price(age, mileage, engine):
    price = 500000 - 20000 * age - 2 * mileage + 500 * engine
    return price
# Example car
age = 3
mileage = 30000
engine = 1500

price = predict_price(age, mileage, engine)

print("Predicted Car Price: ₹", price)
```
