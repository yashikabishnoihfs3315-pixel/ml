Question 1: Comparing Mean, Median, and Mode Imputation

(A) Missing values and imputation

```python

import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    "Age": [22,25,np.nan,30,35,100,np.nan,28],
    "Monthly_Income": [20000,25000,30000,np.nan,
                       35000,500000,28000,np.nan],
    "City": ["Delhi","Mumbai","Delhi",np.nan,
             "Pune","Delhi",np.nan,"Mumbai"],
    "Purchase_Category": ["Electronics","Clothing",np.nan,
                          "Electronics","Clothing","Electronics",
                          "Clothing",np.nan]
})

print("Missing Values:")
print(df.isnull().sum())

num = ["Age", "Monthly_Income"]
cat = ["City", "Purchase_Category"]

# Mean imputation
mean_df = df.copy()
mean_df[num] = SimpleImputer(strategy="mean").fit_transform(df[num])

# Median imputation
median_df = df.copy()
median_df[num] = SimpleImputer(strategy="median").fit_transform(df[num])

# Mode imputation
for col in cat:
    imp = SimpleImputer(strategy="most_frequent")
    mean_df[[col]] = imp.fit_transform(df[[col]])
    median_df[[col]] = imp.fit_transform(df[[col]])

print("\nMean Imputation:")
print(mean_df)

print("\nMedian Imputation:")
print(median_df)
```

(B) Statistical comparison

```python

def stats(data):
    return data[num].agg(["mean","median","std"]).round(2)

print("Original Statistics:")
print(stats(df))

print("\nMean Imputation Statistics:")
print(stats(mean_df))

print("\nMedian Imputation Statistics:")
print(stats(median_df))
```

Question 2: Implementing SimpleImputer with Different Strategies

(A) SimpleImputer with different strategies

```python

import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    "Patient_Age": [25,32,np.nan,45,52,np.nan,38,60],
    "Blood_Sugar": [90,110,105,np.nan,250,130,np.nan,115],
    "Blood_Group": ["A","B","O",np.nan,"A","B",np.nan,"A"],
    "Treatment": ["Basic",np.nan,"Advanced","Basic",
                  "Advanced","Basic",np.nan,"Basic"]
})

df1 = df.copy()

df1[["Patient_Age"]] = SimpleImputer(strategy="mean").fit_transform(df[["Patient_Age"]])
df1[["Blood_Sugar"]] = SimpleImputer(strategy="median").fit_transform(df[["Blood_Sugar"]])

for col in ["Blood_Group", "Treatment"]:
    df1[[col]] = SimpleImputer(strategy="most_frequent").fit_transform(df[[col]])

print("Workflow A:")
print(df1)
print("\nMissing Values:")
print(df1.isnull().sum())
```

(B) Constant imputation

```python

df2 = df.copy()

num = ["Patient_Age", "Blood_Sugar"]
cat = ["Blood_Group", "Treatment"]

df2[num] = SimpleImputer(
    strategy="constant", fill_value=0
).fit_transform(df[num])

df2[cat] = SimpleImputer(
    strategy="constant", fill_value="Unknown"
).fit_transform(df[cat])

print("Workflow B:")
print(df2)
print("\nMissing Values:")
print(df2.isnull().sum())
```

Question 3: Investigating the Impact of Outliers on Imputation

(A) Mean vs Median Imputation

```python

import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

df = pd.DataFrame({
    "Customer_ID": range(1,11),
    "Annual_Spending": [12000,15000,np.nan,18000,22000,
                        25000,30000,np.nan,900000,20000],
    "Order_Count": [4,5,6,np.nan,8,7,9,10,50,np.nan],
    "Membership": ["Silver","Gold","Silver","Gold",np.nan,
                   "Silver","Gold",np.nan,"Platinum","Silver"]
})

num = ["Annual_Spending", "Order_Count"]

mean_df = df.copy()
median_df = df.copy()

mean_df[num] = SimpleImputer(strategy="mean").fit_transform(df[num])
median_df[num] = SimpleImputer(strategy="median").fit_transform(df[num])

print("Mean Imputation:")
print(mean_df[num])

print("\nMedian Imputation:")
print(median_df[num])

print("\nAnnual Spending Statistics:")
print("Mean:", mean_df["Annual_Spending"].mean().round(2))
print("Std:", mean_df["Annual_Spending"].std().round(2))

print("\nMedian Imputation Statistics:")
print("Mean:", median_df["Annual_Spending"].mean().round(2))
print("Std:", median_df["Annual_Spending"].std().round(2))
```

(B) ColumnTransformer + SimpleImputer

```python

from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

num_cols = ["Annual_Spending", "Order_Count"]
cat_cols = ["Membership"]

preprocessor = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), num_cols),
    ("cat", SimpleImputer(strategy="most_frequent"), cat_cols)
])

X = df[num_cols + cat_cols]
result = preprocessor.fit_transform(X)

new_df = pd.DataFrame(result, columns=num_cols + cat_cols)

print("Transformed Dataset:")
print(new_df)

print("\nMissing Values:")
print(new_df.isnull().sum())

print("\nFeature Columns:")
print(new_df.columns.tolist())
```
