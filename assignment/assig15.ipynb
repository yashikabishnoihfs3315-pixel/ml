Question 1: OLS Mathematical Challenge

```python

import numpy as np

x = np.array([1,2,4,5,7,8])
y = np.array([3,4,7,8,12,10])

# OLS
m, c = np.polyfit(x, y, 1)
pred = m*x + c
r = y - pred

print("Mean x:", x.mean())
print("Mean y:", y.mean())
print("Slope:", m, "Intercept:", c)
print("Predicted:", pred.round(2))
print("Residuals:", r.round(2))
print("SSE:", (r**2).sum())
print("MSE:", (r**2).mean())

# Outlier
y[5] = 25
m, c = np.polyfit(x, y, 1)
print("New slope:", m, "New intercept:", c)
```

Question 2: Build a Regression Model from Scratch

```python

import numpy as np

x = np.array([2,4,6,8,10,12], dtype=float)
y = np.array([15,20,28,35,43,48], dtype=float)

# Slope and intercept
m = np.sum((x-x.mean())*(y-y.mean())) / np.sum((x-x.mean())**2)
c = y.mean() - m*x.mean()

# Prediction
pred = m*x + c
r = y - pred

print("Slope:", m, "Intercept:", c)
print("Predictions:", m*np.array([5,9,16]) + c)
print("Residuals:", r.round(2))
print("SSE:", np.sum(r**2))
print("MSE:", np.mean(r**2))
print("RMSE:", np.sqrt(np.mean(r**2)))
```

Question 3: Regression Model Selection and Extrapolation Risk

```python

import numpy as np

hours = np.array([40, 85, 120, 200])
fuel = 25 + 1.8 * hours

for h, f in zip(hours, fuel):
    print(h, "hours:", f, "-", 
          "Interpolation" if h <= 100 else "Extrapolation")
```
