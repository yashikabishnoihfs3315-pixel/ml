Q1. Student Performance DataFrame: Construction and Analysis step by step using Python and Pandas.

1.Create a DataFrame and display records

```python
import pandas as pd

data = {
    "Student_ID": [101, 102, 103, 104, 105, 106,
                   107, 108, 109, 110, 111, 112],
    "Name": ["Amit", "Priya", "Rahul", "Neha",
             "Vikas", "Pooja", "Karan", "Simran",
             "Arjun", "Meena", "Rohit", "Anjali"],
    "Age": [19, 20, 18, 21, 20, 19, 22, 18, 21, 20, 19, 22],
    "Hours_Studied": [2, 5, 3, 8, 6, 4, 9, 2, 7, 5, 6, 8],
    "Attendance": [65, 85, 70, 95, 88, 75, 98, 60, 90, 82, 86, 94],
    "Marks": [48, 78, 55, 92, 84, 67, 96, 45, 89, 76, 81, 93]
}

df = pd.DataFrame(data)

print(df.head())

print(df.tail(3))
```

```python
selected = df[["Name", "Hours_Studied", "Marks"]]

print(selected)

avg_marks = df["Marks"].mean()

print("Average Marks:", avg_marks)

above_avg = selected[selected["Marks"] > avg_marks]

print(above_avg)
```

Step 5. Complete data-preparation pipeline

```python

import pandas as pd
from sklearn.model_selection import train_test_split

def prepare_data(data):
    # 1. Copy DataFrame
    data = data.copy()

    # 2. Define features and target
    features = ["Hours_Studied", "Attendance", "Sleep_Hours"]
    target = "Final_Marks"

    # 3. Validate columns
    required = features + [target]
    missing_cols = [c for c in required if c not in data.columns]

    if missing_cols:
        raise ValueError(f"Missing columns: {missing_cols}")

    # 4. Validate missing values
    if data[required].isnull().any().any():
        raise ValueError("Missing values found!")

    # 5. Create X and y
    X = data[features]
    y = data[target]

    # 6. Split dataset
    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42
    )

    # 7. Report
    print("Total rows:", len(data))
    print("Feature columns:", features)
    print("Target column:", target)
    print("X_train:", X_train.shape)
    print("X_test:", X_test.shape)
    print("y_train:", y_train.shape)
    print("y_test:", y_test.shape)

    print("Indices overlap:",
          not set(X_train.index).isdisjoint(set(X_test.index)))

    return X_train, X_test, y_train, y_test

X_train, X_test, y_train, y_test = prepare_data(df)
```
