Question 1: Correlation and Causation in E-commerce

```python
import pandas as pd
import numpy as np
```

```python
data = {"Customer": [1, 2, 3, 4, 5, 6],
    "Emails": [2, 4, 5, 7, 8, 10],
    "Income": [20000, 30000, 35000, 50000, 60000, 70000],
    "Purchases": [1, 2, 2, 4, 5, 6]}

df = pd.DataFrame(data)

print("E-commerce Dataset:")
print(df)

print("\nCorrelation Matrix:")
print(df[["Emails", "Income", "Purchases"]].corr())
```

Question 3: Correlation Matrix and VIF Calculation

```python
import pandas as pd

data = {"Area_sqft": [1000, 1200, 1500, 1800, 2000, 2200],
    "Area_sqm": [92.9, 111.5, 139.4, 167.2, 185.8, 204.4],
    "Bedrooms": [2, 2, 3, 3, 4, 4],
    "Bathrooms": [1, 2, 2, 3, 3, 4],
    "Floors": [1, 1, 2, 2, 2, 3],
    "Price": [25, 30, 40, 50, 60, 70]}

df = pd.DataFrame(data)

print("House Price Dataset:")
print(df)

# Calculate correlation
corr = df.corr()

print("\nCorrelation Matrix:")
print(corr)

print("\nHighly Correlated Features:")

for i in range(len(corr.columns)):
    for j in range(i):
        if abs(corr.iloc[i, j]) > 0.8:
            print(corr.columns[i], "and", corr.columns[j])
```
