Question 1: Multiple Linear Regression and Coefficient Interpretation

```python

import numpy as np

# Area, bedrooms, age, location score
X = np.array([[1200,2,10,3],
              [1800,3,5,4],
              [2400,4,8,5],
              [1500,3,12,2]])

b = np.array([0.045, 6, -1.2, 8])

price = 12 + X @ b
print("Prices:", price)

# House B with new area coefficient
b[0] = 0.060
print("New House B Price:", 12 + X[1] @ b)
```
