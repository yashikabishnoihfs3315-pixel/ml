Numerical and Categorical Data Using Tips Dataset

1. Load the Tips dataset into a DataFrame

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("tips")

print(df)
```

2. Display the first five rows

```python
print(df.head())
```

3. Check the data types of all columns

```python
print(df.dtypes)
```

4. Filter customers whose total bill is greater than 20

```python
result = df[df["total_bill"] > 20]

print(result)
```

5. Why is size numerical but day categorical?

```python
print("Size data type:", df["size"].dtype)
print("Day data type:", df["day"].dtype)

print("Size: Numerical and Discrete")
print("Day: Categorical and Nominal")
```
