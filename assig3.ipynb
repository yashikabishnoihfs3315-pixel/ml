Q1. Employee Salary Analysis

```python
salaries = [18, 20, 22, 22, 24,25, 25, 26, 28, 30,32, 35, 38, 42, 150]
```

(a) Calculate Mean, Median and All Modes

```python
def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2

    if n % 2 == 0:
        return (sorted_data[mid - 1] +
                sorted_data[mid]) / 2
    else:
        return sorted_data[mid]

def calculate_modes(data):
    frequency = {}

    for x in data:
        frequency[x] = frequency.get(x, 0) + 1

    max_freq = max(frequency.values())

    if max_freq == 1:
        return []

    return [x for x, count in frequency.items()
            if count == max_freq]

print("Mean:", calculate_mean(salaries))
print("Median:", calculate_median(salaries))
print("Modes:", calculate_modes(salaries))
```

(b) Remove salary 150 and recalculate

```python
new_salaries = [s for s in salaries if s != 150]

old_mean = calculate_mean(salaries)
old_median = calculate_median(salaries)

new_mean = calculate_mean(new_salaries)
new_median = calculate_median(new_salaries)

print("New Mean:", new_mean)
print("New Median:", new_median)
print("New Modes:", calculate_modes(new_salaries))

mean_change = ((new_mean - old_mean) /
               old_mean) * 100

median_change = ((new_median - old_median) /
                 old_median) * 100

print("Mean % Change:", mean_change)
print("Median % Change:", median_change)
```

Q2. Customer Spending Analysis Using the Tips Dataset

(a) Calculate Mean, Median and All Modes

```python

import seaborn as sns
import pandas as pd

# Load dataset
tips = sns.load_dataset("tips")

# Function to calculate mean
def calculate_mean(data):
    return sum(data) / len(data)

# Function to calculate median
def calculate_median(data):
    data = sorted(data)
    n = len(data)
    mid = n // 2

    if n % 2 == 0:
        return (data[mid - 1] + data[mid]) / 2
    return data[mid]

# Function to calculate all modes
def calculate_modes(data):
    freq = {}

    for x in data:
        freq[x] = freq.get(x, 0) + 1

    max_freq = max(freq.values())

    if max_freq == 1:
        return []

    return [x for x, count in freq.items()
            if count == max_freq]

# Total bill statistics
print("Total Bill Mean:",
      calculate_mean(tips["total_bill"]))
print("Total Bill Median:",
      calculate_median(tips["total_bill"]))
print("Total Bill Modes:",
      calculate_modes(tips["total_bill"]))

# Tip statistics
print("\nTip Mean:",
      calculate_mean(tips["tip"]))
print("Tip Median:",
      calculate_median(tips["tip"]))
print("Tip Modes:",
      calculate_modes(tips["tip"]))

# Lunch and Dinner analysis
for group in ["Lunch", "Dinner"]:
    bills = tips.loc[tips["time"] == group,
                     "total_bill"].tolist()

    print("\n", group)
    print("Mean:", calculate_mean(bills))
    print("Median:", calculate_median(bills))
```

(b) Exclude the Three Largest Bills

```python

# Original bill data
bills = tips["total_bill"]

# Original statistics
old_mean = calculate_mean(bills)
old_median = calculate_median(bills)

# Find three largest bills
largest_three = tips.nlargest(3, "total_bill")

print("Three Largest Bills:")
print(largest_three[["total_bill", "tip", "time"]])

# Temporarily exclude these records
cleaned_tips = tips.drop(largest_three.index)
new_bills = cleaned_tips["total_bill"]

# New statistics
new_mean = calculate_mean(new_bills)
new_median = calculate_median(new_bills)

print("\nBefore Exclusion")
print("Mean:", old_mean)
print("Median:", old_median)

print("\nAfter Exclusion")
print("Mean:", new_mean)
print("Median:", new_median)

# Percentage changes
mean_change = ((new_mean - old_mean) /
               old_mean) * 100

median_change = ((new_median - old_median) /
                 old_median) * 100

print("\nMean Percentage Change:", mean_change)
print("Median Percentage Change:", median_change)
```

Q3. Build a Central Tendency Calculator From Scratch

(a) Python Program

```python
data_1 = [12, 15, 15, 18, 20, 22, 25]
data_2 = [5, 5, 10, 10, 15, 20]
data_3 = [7, 14, 21, 28, 35]
data_4 = [100, 100, 100, 100]
data_5 = [1, 2, 3, 4, 5, 100]

def mean(data):
    return sum(data) / len(data)

def median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2

    if n % 2 == 0:
        return (sorted_data[mid - 1] +
                sorted_data[mid]) / 2
    else:
        return sorted_data[mid]

def mode(data):
    freq = {}

    for x in data:
        freq[x] = freq.get(x, 0) + 1

    max_freq = max(freq.values())

    if max_freq == 1:
        return "No mode"

    return [x for x, count in freq.items()
            if count == max_freq]

datasets = {
    "Data 1": data_1,
    "Data 2": data_2,
    "Data 3": data_3,
    "Data 4": data_4,
    "Data 5": data_5
}

print("CENTRAL TENDENCY REPORT")
print("-" * 50)

for name, data in datasets.items():
    print("\n", name, ":", data)
    print("Mean  :", round(mean(data), 2))
    print("Median:", median(data))
    print("Mode  :", mode(data))
```

(b) Formatted Report

```python
print("=" * 65)
print("          MEASURES OF CENTRAL TENDENCY REPORT")
print("=" * 65)

print(f"{'Dataset':<12}{'Mean':<12}{'Median':<12}{'Mode':<20}")
print("-" * 65)

for name, data in datasets.items():
    m = mean(data)
    med = median(data)
    mo = mode(data)

    print(f"{name:<12}{m:<12.2f}{med:<12}"
          f"{str(mo):<20}")

print("=" * 65)
```

Q4. Statistical Transformations and Reasoning.

```python
orders = [1, 1, 1, 2, 2, 2, 2, 3, 3, 4,4, 5, 5, 6, 8, 10, 15, 20, 25, 80]
```

(a) Calculate Mean, Median and All Modes

```python
def mean(data):
    return sum(data) / len(data)

def median(data):
    data = sorted(data)
    n = len(data)
    mid = n // 2

    if n % 2 == 0:
        return (data[mid-1] + data[mid]) / 2
    return data[mid]

def mode(data):
    freq = {}
    for x in data:
        freq[x] = freq.get(x, 0) + 1

    max_freq = max(freq.values())

    if max_freq == 1:
        return "No mode"

    return [x for x, c in freq.items()
            if c == max_freq]

# Original measures
print("Original Mean:", mean(orders))
print("Original Median:", median(orders))
print("Original Mode:", mode(orders))

# Remove largest observation
new_orders = orders.copy()
new_orders.remove(max(new_orders))

print("\nAfter Removing Largest Observation")
print("Mean:", mean(new_orders))
print("Median:", median(new_orders))
print("Mode:", mode(new_orders))
```

(b) Statistical Transformations

```python
# Transformation 1: Add 20
add_20 = [x + 20 for x in orders]

# Transformation 2: Multiply by 3
multiply_3 = [x * 3 for x in orders]

# Calculate and compare measures
datasets = {
    "Original": orders,
    "Add 20": add_20,
    "Multiply by 3": multiply_3
}

print("STATISTICAL TRANSFORMATION REPORT")
print("-" * 60)
print(f"{'Dataset':<18}{'Mean':<12}"
      f"{'Median':<12}{'Mode'}")
print("-" * 60)

for name, data in datasets.items():
    print(f"{name:<18}{mean(data):<12.2f}"
          f"{median(data):<12}"
          f"{mode(data)}")

# Verify transformation rules
print("\nVerification:")

print("Mean + 20:", mean(orders) + 20)
print("Median + 20:", median(orders) + 20)
print("Mode + 20:", [m + 20 for m in mode(orders)])

print("\nMean x 3:", mean(orders) * 3)
print("Median x 3:", median(orders) * 3)
print("Mode x 3:", [m * 3 for m in mode(orders)])
```

Q5. Missing Values and Outlier Investigation.

```python
sales = [12, 15, 14, 16, None,15, 18, 14, 17, 16,None, 19, 15, 120, 14]
```

(a) Identify Missing Values, Calculate Central Tendency and Detect Outliers

```python
import pandas as pd
```

```python
s = pd.Series(sales)

missing = s.isna()
missing_count = missing.sum()
missing_percent = (missing_count / len(s)) * 100

clean_sales = s.dropna().tolist()

def mean(data):
    return sum(data) / len(data)

def median(data):
    data = sorted(data)
    n = len(data)
    mid = n // 2

    if n % 2 == 0:
        return (data[mid-1] + data[mid]) / 2
    return data[mid]

def mode(data):
    freq = {}
    for x in data:
        freq[x] = freq.get(x, 0) + 1

    max_freq = max(freq.values())

    if max_freq == 1:
        return "No mode"

    return [x for x, c in freq.items()
            if c == max_freq]

print("Sales Series:")
print(s)

print("\nMissing observations:", missing_count)
print("Missing positions:", list(s.index[missing]))
print("Missing percentage:", missing_percent, "%")

print("\nMean:", mean(clean_sales))
print("Median:", median(clean_sales))
print("Mode:", mode(clean_sales))
```

(b) Recalculate After Excluding the Suspected Outlier

```python
old_mean = mean(clean_sales)
old_median = median(clean_sales)

filtered_sales = [x for x in clean_sales if x != 120]

new_mean = mean(filtered_sales)
new_median = median(filtered_sales)

print("Before Exclusion")
print("Mean:", old_mean)
print("Median:", old_median)
print("Mode:", mode(clean_sales))

print("\nAfter Exclusion")
print("Mean:", new_mean)
print("Median:", new_median)
print("Mode:", mode(filtered_sales))

mean_change = ((new_mean - old_mean) /
               old_mean) * 100

median_change = ((new_median - old_median) /
                 old_median) * 100

print("\nMean Percentage Change:", round(mean_change, 2), "%")
print("Median Percentage Change:", round(median_change, 2), "%")
```
