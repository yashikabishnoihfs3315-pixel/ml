```python
import numpy as np
from sklearn.preprocessing import PowerTransformer
```

```python
# A right-skewed feature
X = np.array([[-1],[0],[3],[5],[10],[50]])

# Create the transformer
pt = PowerTransformer(method="yeo-johnson")

# Fit and transform
X_transformed = pt.fit_transform(X)

print("Original:")
print(X.ravel())

print("\nTransfromed:")
print(np.round(X_transformed.ravel(),3))

print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
# A right-skewed feature
y = np.array([[1],[1],[3],[5],[10],[50]])

# Create the transformer
pt = PowerTransformer(method="box-cox")

# Fit and transform
y_transformed = pt.fit_transform(y)

print("Original:")
print(y.ravel())

print("\nTransfromed:")
print(np.round(y_transformed.ravel(),3))

print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
import numpy as np
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry'],
    'Age': [25, np.nan, 30, 35, np.nan, 40, 28, 33],                  # Numerical with Nulls
    'Salary': [50000, 60000, np.nan, 80000, 65000, np.nan, 52000, 75000], # Numerical with Nulls
    'Experience': [1, 3, 5, np.nan, 4, 10, 2, np.nan],               # Numerical with Nulls
    'Department': ['IT', 'HR', None, 'Finance', 'IT', 'Marketing', None, 'Finance'], # Categorical with Nulls
    'City': ['New York', None, 'Chicago', 'Boston', 'New York', None, 'Chicago', 'Boston'], # Categorical with Nulls
    'Gender': ['F', 'M', 'M', None, 'F', 'M', 'F', None]             # Categorical with Nulls
}

df = pd.DataFrame(data)
df
```
