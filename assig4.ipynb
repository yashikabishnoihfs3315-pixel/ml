Q1. Employee Performance: Population vs. Sample Variance

```python
scores = [45, 50, 52, 55, 58, 60, 65, 75]
```

(a) Python Code

```python
s = pd.Series(scores)

n = len(scores)

mean = sum(scores) / n

squared_deviations = [(x - mean)**2 for x in scores]

pop_variance = sum(squared_deviations) / n

sample_variance = sum(squared_deviations) / (n - 1)

pop_std = pop_variance ** 0.5
sample_std = sample_variance ** 0.5

print("Employee Performance Analysis")
print("-" * 40)
print("Mean:", mean)
print("Population Variance:", pop_variance)
print("Sample Variance:", sample_variance)
print("Population Standard Deviation:", pop_std)
print("Sample Standard Deviation:", sample_std)
```

(b) Why do Population and Sample Variance Differ?

```python
print("NumPy Population Variance:", np.var(scores, ddof=0))
print("NumPy Sample Variance:", np.var(scores, ddof=1))

print("NumPy Population SD:", np.std(scores, ddof=0))
print("NumPy Sample SD:", np.std(scores, ddof=1))
```

Q2. Customer Spending: Comparing Data Spread

```python
import seaborn as sns
import pandas as pd
import numpy as np
```

(a) Calculate Variance, Standard Deviation, Mean and CV

```python
tips = sns.load_dataset("tips")

def mean(data):
    return sum(data) / len(data)

def population_variance(data):
    avg = mean(data)
    return sum((x - avg)**2 for x in data) / len(data)

def population_std(data):
    return population_variance(data)**0.5

for col in ["total_bill", "tip"]:
    data = tips[col].dropna().tolist()

    avg = mean(data)
    var = population_variance(data)
    sd = population_std(data)
    cv = (sd / avg) * 100

    print("\nColumn:", col)
    print("Mean:", round(avg, 4))
    print("Population Variance:", round(var, 4))
    print("Population SD:", round(sd, 4))
    print("CV:", round(cv, 2), "%")
```

(b) Exclude the Five Largest Bills

```python
# Original bill data
bills = tips["total_bill"]

# Find five largest bills
largest_five = tips.nlargest(5, "total_bill")

print("Five Largest Bills:")
print(largest_five[["total_bill", "tip"]])

# Original measures
old_mean = mean(bills.tolist())
old_var = population_variance(bills.tolist())
old_sd = population_std(bills.tolist())

# Exclude the five largest observations
cleaned_tips = tips.drop(largest_five.index)
new_bills = cleaned_tips["total_bill"].tolist()

# New measures
new_mean = mean(new_bills)
new_var = population_variance(new_bills)
new_sd = population_std(new_bills)

print("\nBefore Exclusion")
print("Mean:", old_mean)
print("Variance:", old_var)
print("Standard Deviation:", old_sd)

print("\nAfter Exclusion")
print("Mean:", new_mean)
print("Variance:", new_var)
print("Standard Deviation:", new_sd)

# Percentage changes
mean_change = ((new_mean - old_mean) / old_mean) * 100
var_change = ((new_var - old_var) / old_var) * 100
sd_change = ((new_sd - old_sd) / old_sd) * 100

print("\nPercentage Changes")
print("Mean:", round(mean_change, 2), "%")
print("Variance:", round(var_change, 2), "%")
print("SD:", round(sd_change, 2), "%")
```

Q3. Build a Variance and Standard Deviation Calculator

(a) Python Program

```python
data_1 = [10, 20, 30, 40, 50]
data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]

def calculate_mean(data):
    return sum(data) / len(data)

def population_variance(data):
    avg = calculate_mean(data)
    total = sum((x - avg)**2 for x in data)
    return total / len(data)

def sample_variance(data):
    if len(data) < 2:
        raise ValueError("At least 2 values required")

    avg = calculate_mean(data)
    total = sum((x - avg)**2 for x in data)
    return total / (len(data) - 1)

def population_std(data):
    return population_variance(data)**0.5

def sample_std(data):
    return sample_variance(data)**0.5
```

(b) Formatted Report for All Four Datasets

```python
datasets = {
    "Data 1": data_1,
    "Data 2": data_2,
    "Data 3": data_3,
    "Data 4": data_4
}

print("=" * 100)
print("         VARIANCE AND STANDARD DEVIATION REPORT")
print("=" * 100)

print(f"{'Dataset':<10}{'Mean':<10}{'Pop Var':<12}"
      f"{'Sample Var':<14}{'Pop SD':<12}{'Sample SD':<12}")

print("-" * 100)

for name, data in datasets.items():
    print(f"{name:<10}"
          f"{calculate_mean(data):<10.2f}"
          f"{population_variance(data):<12.2f}"
          f"{sample_variance(data):<14.2f}"
          f"{population_std(data):<12.2f}"
          f"{sample_std(data):<12.2f}")

print("=" * 100)
```

Q4. Investigating the Effect of Mathematical Transformations

```python
measurements = [12, 15, 18, 20, 25, 30, 40]
```

(a) Create Two Transformed Datasets

```python
def calculate_mean(data):
    return sum(data) / len(data)

def calculate_variance(data):
    avg = calculate_mean(data)
    return sum((x - avg)**2 for x in data) / len(data)

def calculate_std(data):
    return calculate_variance(data)**0.5

data_1 = measurements

data_2 = [x + 10 for x in measurements]

data_3 = [x * 4 for x in measurements]

datasets = {
    "Original": data_1,
    "Add 10": data_2,
    "Multiply by 4": data_3
}

print("=" * 70)
print("       MATHEMATICAL TRANSFORMATION REPORT")
print("=" * 70)
print(f"{'Dataset':<18}{'Mean':<12}"
      f"{'Variance':<15}{'Std Dev':<12}")
print("-" * 70)

for name, data in datasets.items():
    print(f"{name:<18}"
          f"{calculate_mean(data):<12.2f}"
          f"{calculate_variance(data):<15.2f}"
          f"{calculate_std(data):<12.2f}")

m = calculate_mean(data_1)
v = calculate_variance(data_1)
sd = calculate_std(data_1)

print("\nVerification:")
print("Original Mean + 10:", m + 10)
print("Transformed Mean:", calculate_mean(data_2))

print("Original Variance:", v)
print("Add 10 Variance:", calculate_variance(data_2))

print("Original Variance x 16:", v * 16)
print("Multiply 4 Variance:", calculate_variance(data_3))

print("Original SD x 4:", sd * 4)
print("Multiply 4 SD:", calculate_std(data_3))
```

(b) Multiply Every Observation by −2

```python
data_4 = [x * (-2) for x in measurements]

print("New Dataset:", data_4)
print("Mean:", calculate_mean(data_4))
print("Variance:", calculate_variance(data_4))
print("Standard Deviation:", calculate_std(data_4))

print("\nVerification:")
print("Expected Mean:", calculate_mean(measurements) * (-2))
print("Expected Variance:", calculate_variance(measurements) * 4)
print("Expected SD:", calculate_std(measurements) * 2)
```

Q5. Outlier Detection and Robust Spread Analysis

```python
sales = [12, 14, 15, 15, 16, 17, 18,19, 20, 21, 22, 24, 25, 120]
```

(a) Python code

```python
mean = sum(sales) / len(sales)

def find_median(data):
    data = sorted(data)
    n = len(data)
    if n % 2 == 0:
        return (data[n//2 - 1] + data[n//2]) / 2
    else:
        return data[n//2]

median = find_median(sales)

variance = sum((x - mean)**2 for x in sales) / len(sales)

sd = variance ** 0.5

data = sorted(sales)
n = len(data)

lower_half = data[:n//2]
upper_half = data[n//2:]

Q1 = find_median(lower_half)
Q3 = find_median(upper_half)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in sales
            if x < lower_bound or x > upper_bound]

print("Mean:", mean)
print("Median:", median)
print("Population Variance:", variance)
print("Population SD:", sd)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

(b) Exclude the flagged outlier and recalculate

```python
clean_sales = [x for x in sales if x not in outliers]

new_mean = sum(clean_sales) / len(clean_sales)
new_median = find_median(clean_sales)

new_variance = sum(
    (x - new_mean)**2 for x in clean_sales
) / len(clean_sales)

new_sd = new_variance ** 0.5

variance_change = (
    (variance - new_variance) / variance
) * 100

sd_change = ( (sd - new_sd) / sd) * 100

print("Clean Sales:", clean_sales)
print("New Mean:", new_mean)
print("New Median:", new_median)
print("New Variance:", new_variance)
print("New SD:", new_sd)
print("Variance Decrease (%):", variance_change)
print("SD Decrease (%):", sd_change)
```
